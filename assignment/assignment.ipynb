### Connections

```python
import os

from dotenv import load_dotenv
from pymongo.mongo_client import MongoClient
from pymongo.server_api import ServerApi
```

```python
# Load environment variables from .env file
load_dotenv()
MONGODB_URI = os.getenv("MONGODB_URI")
if not MONGODB_URI:
    raise ValueError(
        "❌ MONGODB_URI not found!\n"
        "Please create a .env file with your MongoDB credentials.\n"
        "See README.md for setup instructions."
    )
client = MongoClient(MONGODB_URI, server_api=ServerApi("1"))
# Send a ping to confirm a successful connection
try:
    client.admin.command("ping")
    print("✅ Successfully connected to MongoDB!")
except Exception as e:
    print(e)
```

```python
db = client.sample_mflix
movies = db.movies
print(f"📊 Database: {db.name}")
print(f"📁 Collection: movies ({movies.count_documents({})} documents)")
```

```python
#print the structure of the first document in the movies collection
first_movie = movies.find_one()
print("\n🎬 Sample Movie Document:")
for key, value in first_movie.items():
    print(f"  {key}: {value}")
```


### Question 1

Question: From the `movies` collection, return the documents with the `plot` that starts with `"war"` in acending order of released date, print only title, plot and released fields. Limit the result to 5.


**Answer:**

```python
# from movies return the docyments with plot that starts with "war" in ascending order of the released date and print the title, plot and released fields limited to 5 results
query = {"plot": {"$regex": "^war", "$options": "i"}}
projection = {"_id": 0, "title": 1, "plot": 1, "released": 1}
results = movies.find(query, projection).sort("released", 1).limit(5)
print("\n🎬 Movies with plot starting with 'war':")
for movie in results:
    print(f"Title: {movie['title']}, Plot: {movie['plot']}, Released: {movie['released']}")
```

### Question 2

Question: Group by `rated` and count the number of movies in each.

**Answer:**

```python
#group by rated and count the number of movies in each
pipeline = [
    {"$group": {"_id": "$rated", "count": {"$sum": 1}}},
    {"$sort": {"count": -1}}
]
results = movies.aggregate(pipeline)
print("\n📊 Movie counts by rating:")
for result in results:
    print(f"Rating: {result['_id']}, Count: {result['count']}")
```


### Question 3

Question: Count the number of movies with 3 comments or more.


```python
#print the first 3 documents in the movies collection and their comments
print("\n🎬 Sample Movies:")
for i, movie in enumerate(movies.find().limit(3)):
    print(f"  {i+1}. {movie['title']}")
    print("     Comments:")
    for comment in movie.get('comments', []):
        print(f"       - {comment}")    
```

**Answer:**


```python
# ============================================================================
# Question 3
# ============================================================================
#Count the number of movies with 3 comments or more

pipeline = [
    {"$group": {"_id": "$title", "comments": {"$sum": 1}}},
    {"$match": {"comments": {"$gte": 3}}},      
    {"$sort": {"comments": -1}}
]
results = movies.aggregate(pipeline)
print("\n📊 Movie and their comment counts:")
for result in results:
    print(f"Movie: {result['_id']}, Comments: {result['comments']}")
```
